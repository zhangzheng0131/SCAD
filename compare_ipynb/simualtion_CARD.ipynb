## 2. single cell RNAseq  dat

```r
mat <- readMM("data/sc_count_matrix_MB.mtx")
meta_data <- read.table('data/sc_label_louvine_MB.txt',sep='\t',header = TRUE,row.names = 1)
meta_data$barcode=rownames(meta_data)
meta_data$cell_type <- as.character(meta_data$cell_type)
sc_count <- as(mat, "dgCMatrix")
gene_df = read.table('data/sc_gene_MB.txt',sep='\t',header = TRUE,row.names = 1)
rownames(sc_count) <- meta_data$barcode
colnames(sc_count) <- rownames(gene_df)   
```

```r
sc_count=t(sc_count)
colnames(meta_data) <- c("cellType",'sampleInfo')
sc_meta=meta_data
sc_meta$sampleInfo = "sample1"
```

```r
library(CARD)
```

```r
CARD_obj = createCARDObject(
	sc_count = sc_count,
	sc_meta = sc_meta,
	spatial_count = spatial_count,
	spatial_location = spatial_location,
	ct.varname = "cellType",
	ct.select = unique(sc_meta$cellType),
	sample.varname = "sampleInfo",
	minCountGene = 100,
	minCountSpot = 5) 
## QC on scRNASeq dataset! ...
## QC on spatially-resolved dataset! ..
```

```r
library(MuSiC)
```

```r
CARD_obj = CARD_deconvolution(CARD_object = CARD_obj)
```

```r
CARD_obj@Proportion_CARD
write.table(CARD_obj@Proportion_CARD,file = 'results/card_result_propostion.txt',sep='\t',quote = FALSE)
```
